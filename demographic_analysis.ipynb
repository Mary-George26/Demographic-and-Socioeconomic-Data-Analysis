# Outlier Detection

17.	Using the IQR method, investigate potential outliers in the numerical columns:

•	Age
•	Number_of_Dependents
•	Work_Experience
•	Household_Size
•	Income

For every column, calculate and display:

•	Q1
•	Q3
•	IQR
•	Lower bound
•	Upper bound
•	Number of potential outliers


```python
numerical_cols = ['Age', 'Number_of_Dependents', 'Work_Experience', 'Household_Size', 'Income']

for col in numerical_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    print(f"\n---------------------")
    print(f"Column: {col}")
    print(f"Q1: {Q1}")
    print(f"Q3: {Q3}")
    print(f"IQR: {IQR}")
    print(f"Lower bound: {lower_bound}")
    print(f"Upper bound: {upper_bound}")
    print(f"Number of potential outliers: {len(outliers)}\n")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df['Income'])
plt.title('Income Outliers Box Plot')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
df.groupby('Gender')['Income'].mean().plot(kind='bar', color=['skyblue', 'salmon'])
plt.title('Average Income by Gender')
plt.ylabel('Income')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
df.groupby('Primary_Mode_of_Transportation')['Income'].mean().sort_values().plot(kind='barh', color='teal')
plt.title('Average Income by Mode of Transportation')
plt.xlabel('Income')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df[['Age', 'Number_of_Dependents', 'Work_Experience', 'Household_Size', 'Income']].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```
